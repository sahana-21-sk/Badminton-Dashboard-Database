```python
import pandas as pd
import numpy as np

df = pd.read_csv("new_landmarks.csv")

# --------------------------------------------------
# HIP CENTER
# --------------------------------------------------

df["hip_center_x"] = (df["left_hip_x"] + df["right_hip_x"]) / 2
df["hip_center_y"] = (df["left_hip_y"] + df["right_hip_y"]) / 2

# --------------------------------------------------
# MOVEMENT
# --------------------------------------------------

df["dx"] = df["hip_center_x"].diff().fillna(0)
df["dy"] = df["hip_center_y"].diff().fillna(0)

df["distance"] = np.sqrt(
    df["dx"]**2 +
    df["dy"]**2
)

df["speed"] = df["distance"]

df["speed_smoothed"] = (
    df["speed"]
    .rolling(5, min_periods=1)
    .mean()
)

df["acceleration"] = (
    df["speed"]
    .diff()
    .fillna(0)
)

# --------------------------------------------------
# CENTROIDS
# --------------------------------------------------

df["centroid_x"] = (
    df["left_hip_x"] +
    df["right_hip_x"] +
    df["left_shoulder_x"] +
    df["right_shoulder_x"]
) / 4

df["centroid_y"] = (
    df["left_hip_y"] +
    df["right_hip_y"] +
    df["left_shoulder_y"] +
    df["right_shoulder_y"]
) / 4

df["centroid_x_smooth"] = (
    df["centroid_x"]
    .rolling(5, min_periods=1)
    .mean()
)

df["centroid_y_smooth"] = (
    df["centroid_y"]
    .rolling(5, min_periods=1)
    .mean()
)

# --------------------------------------------------
# STANCE
# --------------------------------------------------

df["stance_width"] = np.sqrt(
    (df["left_ankle_x"] - df["right_ankle_x"])**2 +
    (df["left_ankle_y"] - df["right_ankle_y"])**2
)

# --------------------------------------------------
# DISTANCES
# --------------------------------------------------

df["cumulative_distance"] = df["distance"].cumsum()

start_x = df["hip_center_x"].iloc[0]
start_y = df["hip_center_y"].iloc[0]

df["straight_distance"] = np.sqrt(
    (df["hip_center_x"] - start_x)**2 +
    (df["hip_center_y"] - start_y)**2
)

df["path_length"] = df["cumulative_distance"]

df["path_efficiency"] = (
    df["straight_distance"] /
    (df["path_length"] + 1e-6)
)

# --------------------------------------------------
# DIRECTION
# --------------------------------------------------

def get_direction(dx, dy):

    if abs(dx) < 0.002 and abs(dy) < 0.002:
        return "STILL"

    if abs(dx) > abs(dy):
        return "RIGHT" if dx > 0 else "LEFT"

    return "FORWARD" if dy < 0 else "BACKWARD"

df["direction"] = [
    get_direction(dx, dy)
    for dx, dy in zip(df["dx"], df["dy"])
]

direction_map = {
    "STILL":0,
    "LEFT":1,
    "RIGHT":2,
    "FORWARD":3,
    "BACKWARD":4
}

df["direction_code"] = (
    df["direction"]
    .map(direction_map)
)

# --------------------------------------------------
# ONE HOT
# --------------------------------------------------

for d in [
    "BACKWARD",
    "FORWARD",
    "LEFT",
    "RIGHT",
    "STILL"
]:
    df[f"direction_{d}"] = (
        df["direction"] == d
    )

# --------------------------------------------------
# RECOVERY
# --------------------------------------------------

df["recovery_distance"] = (
    df["distance"]
    .rolling(10, min_periods=1)
    .sum()
)

df["recovery_time"] = (
    (df["speed"] < df["speed"].median())
    .astype(int)
)

# --------------------------------------------------
# MOVEMENT RANGE
# --------------------------------------------------

df["movement_range_x"] = (
    df["hip_center_x"].cummax() -
    df["hip_center_x"].cummin()
)

df["movement_range_y"] = (
    df["hip_center_y"].cummax() -
    df["hip_center_y"].cummin()
)

# --------------------------------------------------
# ROLLING
# --------------------------------------------------

df["rolling_speed_mean"] = (
    df["speed"]
    .rolling(10, min_periods=1)
    .mean()
)

df["rolling_speed_std"] = (
    df["speed"]
    .rolling(10, min_periods=1)
    .std()
    .fillna(0)
)

# --------------------------------------------------
# ANGLES
# --------------------------------------------------

df["trajectory_angle"] = np.degrees(
    np.arctan2(
        df["dy"],
        df["dx"]
    )
)

df["direction_change"] = (
    df["trajectory_angle"]
    .diff()
    .fillna(0)
)

# --------------------------------------------------
# COURT ZONE
# --------------------------------------------------

def get_zone(x,y):

    x = np.clip(x,0,0.99)
    y = np.clip(y,0,0.99)

    x_zone = int(x//0.33)
    y_zone = int(y//0.33)

    return x_zone + (y_zone*3)

df["court_zone"] = [
    get_zone(x,y)
    for x,y in zip(
        df["hip_center_x"],
        df["hip_center_y"]
    )
]

# --------------------------------------------------
# STABILITY
# --------------------------------------------------

df["stability_index"] = (
    df["stance_width"] /
    (df["speed"] + 0.1)
)

# --------------------------------------------------

df.to_csv("rf_input.csv", index=False)

print(df.shape)
print("Saved rf_input.csv")
```

```python
import pandas as pd
import joblib

# load model
rf = joblib.load("Badminton_RF_Model (1).pkl")

# load features
df = pd.read_csv("rf_input.csv")

# keep ONLY columns RF expects
X = df[rf.feature_names_in_]

print("X shape:", X.shape)
print("Expected:", len(rf.feature_names_in_))

# predict
preds = rf.predict(X)

# probabilities (if classifier supports it)
try:
    probs = rf.predict_proba(X)
    confidence = probs.max(axis=1)
except:
    confidence = [None] * len(preds)

# save
out = pd.DataFrame({
    "frame": range(len(preds)),
    "prediction": preds,
    "confidence": confidence
})

out.to_csv("predictions.csv", index=False)

print("Predictions shape:", out.shape)
print(out.head())
print("Saved predictions.csv")
```

```python
import cv2
import pandas as pd

video_path = "new_video.mp4"

preds = pd.read_csv("predictions.csv")

cap = cv2.VideoCapture(video_path)

fps = cap.get(cv2.CAP_PROP_FPS)
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))

fourcc = cv2.VideoWriter_fourcc(*'mp4v')

out = cv2.VideoWriter(
    "annotated_video.mp4",
    fourcc,
    fps,
    (width, height)
)

frame_idx = 0

while True:

    ret, frame = cap.read()

    if not ret:
        break

    if frame_idx < len(preds):

        label = str(preds.loc[frame_idx, "prediction"])

        conf = preds.loc[frame_idx, "confidence"]

        text = f"{label} ({conf:.2f})"

        cv2.putText(
            frame,
            text,
            (30, 50),
            cv2.FONT_HERSHEY_SIMPLEX,
            1,
            (0, 255, 0),
            2
        )

        cv2.putText(
            frame,
            f"Frame: {frame_idx}",
            (30, 100),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.8,
            (255, 255, 255),
            2
        )

    out.write(frame)

    frame_idx += 1

cap.release()
out.release()

print("Saved annotated_video.mp4")
print("Frames written:", frame_idx)
```

```python
import pandas as pd
import json

FPS = 60.002

preds = pd.read_csv("predictions.csv")
segments = pd.read_csv("action_segments.csv")

total_frames = len(preds)
duration_seconds = total_frames / FPS

# Action counts
action_counts = (
    preds["prediction"]
    .value_counts()
    .to_dict()
)

# Percentages
action_percentages = {
    k: round(v * 100 / total_frames, 2)
    for k, v in action_counts.items()
}

# Segment information
segment_list = []

for _, row in segments.iterrows():

    start_frame = int(row["start_frame"])
    end_frame = int(row["end_frame"])

    duration_frames = end_frame - start_frame + 1

    segment_list.append({
        "action": row["action"],
        "start_frame": start_frame,
        "end_frame": end_frame,
        "duration_frames": duration_frames,
        "duration_seconds": round(
            duration_frames / FPS,
            2
        )
    })

analytics = {
    "total_frames": total_frames,
    "fps": FPS,
    "duration_seconds": round(
        duration_seconds,
        2
    ),
    "action_counts": action_counts,
    "action_percentages": action_percentages,
    "num_segments": len(segment_list),
    "segments": segment_list
}

with open(
    "analytics.json",
    "w"
) as f:
    json.dump(
        analytics,
        f,
        indent=4
    )

print("Saved analytics.json")
```
