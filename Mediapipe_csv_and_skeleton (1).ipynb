```python
import cv2
import mediapipe as mp
import os
import pandas as pd
from google.colab import drive

print(mp.__version__)
```

```python
input_folder = "/content/drive/MyDrive/Frames of 5/open video 1 bb"

output_folder = "/content/drive/MyDrive/Frames of 5/open video 1 mediapipe"

os.makedirs(output_folder, exist_ok=True)

csv_data = []
```

```python
mp_pose = mp.solutions.pose

mp_drawing = mp.solutions.drawing_utils

pose = mp_pose.Pose(
    static_image_mode=False,
    model_complexity=1,
    smooth_landmarks=True,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
)

print("MediaPipe Ready 🔥")
```

```python
image_files = sorted(os.listdir(input_folder))

print("Total Images:", len(image_files))

for image_name in image_files:

    image_path = os.path.join(input_folder, image_name)

    img = cv2.imread(image_path)

    if img is None:
        continue

    h, w, _ = img.shape

    # =========================
    # CREATE EXPANDED CROP
    # =========================

    padding = 100

    x1 = padding
    y1 = padding

    x2 = w - padding
    y2 = h - padding

    crop = img[y1:y2, x1:x2]

    # Convert to RGB
    crop_rgb = cv2.cvtColor(crop, cv2.COLOR_BGR2RGB)

    # Run MediaPipe
    results = pose.process(crop_rgb)

    # =========================
    # DRAW + SAVE LANDMARKS
    # =========================

    if results.pose_landmarks:

        # Draw skeleton on crop
        mp_drawing.draw_landmarks(
            crop_rgb,
            results.pose_landmarks,
            mp_pose.POSE_CONNECTIONS
        )

        # Save landmark coordinates
        frame_data = {}

        frame_data["frame"] = image_name

        for idx, landmark in enumerate(results.pose_landmarks.landmark):

            # Convert crop coords back to original image coords
            real_x = (landmark.x * (x2 - x1)) + x1
            real_y = (landmark.y * (y2 - y1)) + y1

            frame_data[f"x_{idx}"] = real_x
            frame_data[f"y_{idx}"] = real_y
            frame_data[f"z_{idx}"] = landmark.z
            frame_data[f"visibility_{idx}"] = landmark.visibility

        csv_data.append(frame_data)

        # Put processed crop back into image
        crop_bgr = cv2.cvtColor(crop_rgb, cv2.COLOR_RGB2BGR)

        img[y1:y2, x1:x2] = crop_bgr

    # =========================
    # SAVE FINAL IMAGE
    # =========================

    output_path = os.path.join(output_folder, image_name)

    cv2.imwrite(output_path, img)

print("ALL IMAGES COMPLETED SUCCESSFULLY 🔥")
```

```python
df = pd.DataFrame(csv_data)

csv_path = "/content/drive/MyDrive/Frames of 5/open_video_mp_1_landmarks.csv"

df.to_csv(csv_path, index=False)

print("CSV SAVED SUCCESSFULLY 🔥")
```

```python
import cv2
import mediapipe as mp
import os
import pandas as pd
from google.colab import drive

print(mp.__version__)
```

```python
input_folder = "/content/drive/MyDrive/Frames of 5/open video 2 bb"

output_folder = "/content/drive/MyDrive/Frames of 5/open video 2 mediapipe"

os.makedirs(output_folder, exist_ok=True)
```

```python
mp_pose = mp.solutions.pose

mp_drawing = mp.solutions.drawing_utils

pose = mp_pose.Pose(
    static_image_mode=False,
    model_complexity=2,
    smooth_landmarks=True,
    min_detection_confidence=0.3,
    min_tracking_confidence=0.3
)

print("MediaPipe Ready 🔥")
```

```python
image_files = sorted(os.listdir(input_folder))

csv_data = []

previous_landmarks = None

print("Total Images:", len(image_files))

for image_name in image_files:

    image_path = os.path.join(input_folder, image_name)

    img = cv2.imread(image_path)

    if img is None:
        continue

    img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

    # Run MediaPipe
    results = pose.process(img_rgb)

    current_landmarks = None

    # =========================
    # IF DETECTED
    # =========================

    if results.pose_landmarks:

        current_landmarks = results.pose_landmarks.landmark

        previous_landmarks = current_landmarks

    # =========================
    # FALLBACK TO PREVIOUS
    # =========================

    elif previous_landmarks is not None:

        current_landmarks = previous_landmarks

    # =========================
    # DRAW LANDMARKS
    # =========================

    if current_landmarks is not None:

        h, w, _ = img.shape

        frame_data = {}

        frame_data["frame"] = image_name

        # Draw manually
        for connection in mp_pose.POSE_CONNECTIONS:

            start_idx = connection[0]
            end_idx = connection[1]

            x1 = int(current_landmarks[start_idx].x * w)
            y1 = int(current_landmarks[start_idx].y * h)

            x2 = int(current_landmarks[end_idx].x * w)
            y2 = int(current_landmarks[end_idx].y * h)

            cv2.line(img, (x1, y1), (x2, y2), (0,255,0), 2)

        # Draw joints
        for idx, landmark in enumerate(current_landmarks):

            x = int(landmark.x * w)
            y = int(landmark.y * h)

            cv2.circle(img, (x,y), 4, (0,0,255), -1)

            frame_data[f"x_{idx}"] = landmark.x
            frame_data[f"y_{idx}"] = landmark.y
            frame_data[f"z_{idx}"] = landmark.z
            frame_data[f"visibility_{idx}"] = landmark.visibility

        csv_data.append(frame_data)

    # Save image
    output_path = os.path.join(output_folder, image_name)

    cv2.imwrite(output_path, img)

print("ALL IMAGES COMPLETED SUCCESSFULLY 🔥")
```

```python
df = pd.DataFrame(csv_data)

csv_path = "/content/drive/MyDrive/Frames of 5/open_video_trim_2_landmarks.csv"

df.to_csv(csv_path, index=False)

print("CSV SAVED SUCCESSFULLY 🔥")
```
