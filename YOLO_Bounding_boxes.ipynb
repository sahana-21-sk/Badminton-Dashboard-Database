```python
zip_files = [
    "open video trim 1 frames.zip",
    "open video trim2.zip",
    "open video trim 3.zip",
    "lin dan frames.zip",
    "Badminton Full court.zip"
]

for zip_name in zip_files:

    print(f"\nProcessing {zip_name}...")

    # Full zip path
    zip_path = os.path.join(main_folder, zip_name)

    # Temporary extraction folder
    temp_extract = "/content/temp_frames"

    # Create temp folder
    os.makedirs(temp_extract, exist_ok=True)

    # Unzip files
    with zipfile.ZipFile(zip_path, 'r') as zip_ref:
        zip_ref.extractall(temp_extract)

    # Create output folder name
    folder_name = zip_name.replace(".zip", "") + " frames"

    output_folder = os.path.join(main_folder, folder_name)

    os.makedirs(output_folder, exist_ok=True)

    # Get all frame files
    frame_files = sorted(os.listdir(temp_extract))

    # Process every frame
    for frame_name in frame_files:

        image_path = os.path.join(temp_extract, frame_name)

        # Read image
        img = cv2.imread(image_path)

        if img is None:
            continue

        # YOLO detection
        results = model(image_path)

        boxes = results[0].boxes

        largest_area = 0
        best_box = None
        best_conf = 0

        # Find largest PERSON box
        for box in boxes:

            cls = int(box.cls[0])

            # Person class
            if cls == 0:

                x1, y1, x2, y2 = map(int, box.xyxy[0])

                area = (x2 - x1) * (y2 - y1)

                if area > largest_area:
                    largest_area = area
                    best_box = (x1, y1, x2, y2)
                    best_conf = float(box.conf[0])

        # Draw only best box
        if best_box:

            x1, y1, x2, y2 = best_box

            # Small padding
            padding = 15

            x1 = max(0, x1 - padding)
            y1 = max(0, y1 - padding)
            x2 = min(img.shape[1], x2 + padding)
            y2 = min(img.shape[0], y2 + padding)

            # Draw rectangle
            cv2.rectangle(
                img,
                (x1, y1),
                (x2, y2),
                (255, 0, 0),
                3
            )

            # Confidence label
            label = f"player {best_conf:.2f}"

            cv2.putText(
                img,
                label,
                (x1, y1 - 10),
                cv2.FONT_HERSHEY_SIMPLEX,
                1,
                (255, 0, 0),
                2
            )

        # Save processed image
        save_path = os.path.join(output_folder, frame_name)

        cv2.imwrite(save_path, img)

    print(f"{zip_name} completed!")

    # Clear temp folder
    import shutil

# Clear temp folder properly
for file in os.listdir(temp_extract):

    file_path = os.path.join(temp_extract, file)

    # Remove file
    if os.path.isfile(file_path):
        os.remove(file_path)

    # Remove folder
    elif os.path.isdir(file_path):
        shutil.rmtree(file_path)


print("\nALL VIDEOS COMPLETED SUCCESSFULLY 🔥")
```
